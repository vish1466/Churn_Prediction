```python
# Formatting necessary column data types to numeric dtypes

df.select_dtypes(include = ["str"]).nunique()
```

```python
df.columns
```

```python
df['gender'] = df['gender'].map({'Female' : 0, 'Male' : 1})

binary_col = ["Partner", "Dependents", "PhoneService", "PaperlessBilling", "Churn"]

for col in binary_col:
    df[col] = df[col].map({"Yes" : 1, "No" : 0})

```

```python
df.head()
```

```python
df["TotalCharges"] = pd.to_numeric(df['TotalCharges'], errors='coerce')
print("Number of Empty Charges  : ", df["TotalCharges"].isna().sum())
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
df.info()
```

```python
df = df.drop(columns=["customerID"])
df.head()
```

```python
categorical_cols = df.select_dtypes(include = ["str", "object", "category"]).columns

for col in categorical_cols:
    print(f"__ {col}  ({df[col].nunique()} unique) __")
    print(df[col].unique())
    print()
```

```python
service_cols = ['MultipleLines','OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport',
                'StreamingTV', 'StreamingMovies']


df[service_cols] = df[service_cols].replace({"No internet service" : "No", "No phone service" : "No"})

for col in service_cols:
    df[col] = df[col].map({"Yes" : 1, "No" : 0})

df.info()
```

```python
remaining_cols = ["InternetService", "Contract", "PaymentMethod"]

df = pd.get_dummies(df, columns=remaining_cols,drop_first=True, dtype = int)
df.info()
```
