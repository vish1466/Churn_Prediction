```python
import numpy as np
import pandas as pd

X = pd.read_csv(r"C:\Users\dhanu\OneDrive\Desktop\data_science_project\3_Feature_Eng & Data_Prep\Input_Output_Saved\X_INPUT_DATA")
Y = pd.read_csv(r"C:\Users\dhanu\OneDrive\Desktop\data_science_project\3_Feature_Eng & Data_Prep\Input_Output_Saved\Y_OUTPUT_LABELS")
```

```python
# splitting into training and test datasets

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42, stratify = Y)
```

```python
X.columns
```

```python
# We have some continuous columns, which we have to scale for the training input dataset

scaler = StandardScaler()
cols_to_scale = ['tenure', 'MonthlyCharges', 'TotalCharges_log', 'CostperService', 'AvgMonthlyCharges']

x_train[cols_to_scale] = scaler.fit_transform(x_train[cols_to_scale])
x_test[cols_to_scale] = scaler.transform(x_test[cols_to_scale])

print(f"x_train shape : {x_train.shape} || x_test shape : {x_test.shape}")
```

The dataset (training and test) are both prepared and properly scaled, and ready for model training.

```python
x_train.to_csv(r"C:\Users\dhanu\OneDrive\Desktop\data_science_project\3_Feature_Eng & Data_Prep\Training_and_Test_datasets\x_train", index = False)
x_test.to_csv(r"C:\Users\dhanu\OneDrive\Desktop\data_science_project\3_Feature_Eng & Data_Prep\Training_and_Test_datasets\x_test", index = False)
y_train.to_csv(r"C:\Users\dhanu\OneDrive\Desktop\data_science_project\3_Feature_Eng & Data_Prep\Training_and_Test_datasets\y_train", index = False)
y_test.to_csv(r"C:\Users\dhanu\OneDrive\Desktop\data_science_project\3_Feature_Eng & Data_Prep\Training_and_Test_datasets\y_test", index = False)
```
